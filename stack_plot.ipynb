```python
import matplotlib.pyplot as plt
days = [1,2,3,4,5,6,7]

coding = [2,3,6,4,7,9,1]
gaming =[3,3,1,4,7,8,6]
sleeping = [3,5,6,7,1,4,8]
making_videos = [8,2,1,5,7,4,6]

labels = ["coding","gaming","sleeping","making_videos"]
colors = ['skyblue', 'lightgreen', 'gold', 'lightcoral']

plt.stackplot(days,coding,gaming,sleeping,making_videos,labels=labels,colors=colors)
plt.legend(loc="upper left")
plt.grid(True)
plt.style.use("fast")
plt.xlabel("Days")
plt.ylabel("Hours of day")
plt.title("My Daily Schedule")
```
