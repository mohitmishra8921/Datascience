```python
import matplotlib.pyplot as plt
fig, axs = plt.subplots(2, 2, figsize=(8, 6))
x = [1, 2, 3, 4, 5]
y1 = [i * 2 for i in x]
y2 = [i ** 2 for i in x]
y3 = [i ** 0.5 for i in x]
y4 = [10 - i for i in x]
ys = [y1, y2, y3, y4]
titles = ['x * 2', 'x squared', 'sqrt(x)', '10 - x']

for i in range(2):
    for j in range(2):
        idx = i * 2 + j
        axs[i, j].plot(x, ys[idx])
        axs[i, j].set_title(titles[idx])
        plt.suptitle("Different Patterns Comparision")

plt.tight_layout()
plt.show()
plt.suptitle("Different Patterns Comparision")
```

```python
import matplotlib.pyplot as plt
x = [1,2,4,5,7,9]
y1 = [i*2 for i in x]
y2 = [10-i for i in x]
y3 = [i ** 2 for i in x]
y4 = [i ** 0.5 for i in x]
# For first graph
plt.subplot(2,2,1)
plt.hist(x,y1,edgecolor = "black")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("X*2")
plt.grid()

# For second graph
plt.subplot(2,2,2)
plt.plot(x,y2)
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Root of x")
plt.grid()

# For third graph
plt.subplot(2,2,3)
plt.scatter(x,y3)
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Multiples of 2")
plt.grid()

# For fourth graph
plt.subplot(2,2,4)
plt.stackplot(x,y4)
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Multiplied by 0.5")
plt.grid()
plt.suptitle("Representation of data in various forms of graph")

plt.tight_layout()
```
