```python
import matplotlib.pyplot as plt
import numpy as np

years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
sachin = [98,23,45,14,89,67,37,90,78,98]
virat = [95,123,85,14,9,46,37,78,87,56]
rohit = [90,56,78,98,78,56,90,55,67,87]

x = np.arange(len(years))
width = 0.25

plt.bar(x-width,sachin,width = width,label ="Sachin")
plt.bar(x,virat,width = width,label ="Virat")
plt.bar(x+width,rohit,width = width,label ="Rohit")
plt.xticks(x,years)
plt.ylabel("Runs")
plt.legend()
plt.tight_layout()
```
