```python
import matplotlib.pyplot as plt

years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0]
plt.plot(years, kohli, color='orange', linestyle='--', label="Kohli")
plt.plot(years, sehwag, color='green', linestyle='-.', label="Sehwag")
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Performance Comparison")
plt.legend()
plt.show()
```

```python

labels = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [90099, 50009, 92000, 19500]
colors = ['#ff9989','#66b3ff','#99ff68','#ffcc99']
explode = [0.1,0,0,0]
plt.tight_layout()
plt.title("Player's comparison")
plt.pie(runs, labels=labels, colors=colors,startangle=180,shadow = True, explode = explode,autopct='%1.1f%%',wedgeprops={'edgecolor': 'black','linestyle':'--','linewidth':1.5})
plt.show()
```
