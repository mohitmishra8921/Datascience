```python
import matplotlib.pyplot as plt 
colors = ['blue','red','green','yellow']
languages = ["Python","Java ","C++","Ruby"]
usage = [50,40,20,10]
explode =[0,0,0.1,0]
plt.style.use("ggplot")
plt.pie(usage,labels=languages,explode = explode,colors = colors,shadow = True,wedgeprops={'edgecolor': 'black', 'linewidth': 1, 'linestyle': '-'},autopct='%1.1f%%')
plt.title("Languages Comparison")
plt.show()
```

```python
import matplotlib.pyplot as plt 
activities = ["Studying","Playing","Sleeping","Watching TV"]
time_spent = [3,5,2,4]
plt.figure(figsize = (6,6))
explode = [0,0.1,0,0]
colors = ['blue','yellow','pink','brown']
plt.title("Daily Activity")
plt.pie(time_spent,labels=activities,colors= colors,explode=explode,shadow = True,autopct='%1.1f%%',startangle=250)
print(plt)
plt.style.use('ggplot')
plt.axis('equal')
```
