```python
import matplotlib.pyplot as plt 

years = [1991,1992,1993,1994,1995,1996,1997,1998,1999,2000]
sachin = [0,101,151,200,153,98,57,84,99,100]
kohli =  [99,10,265,134,53,90,99,88,99,43]
rohit = [33,164,32,89,97,145,65,90,199,12]

plt.plot(years,kohli,color = "red",linestyle = '-.',label ="Virat Kohli")
plt.plot(years,sachin,color = "blue",linestyle = ':',label ="Sachin Tendulkar")
plt.plot(years,rohit,color = "purple",linestyle = '--',label ="Rohit Sharma")
plt.legend()
plt.grid(True)
plt.xlabel("Years➡️")
plt.ylabel("Runs➡️")
plt.title("Comparison of cricketers")
plt.style.use("dark_background")
```
