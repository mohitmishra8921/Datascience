```python
import pandas as pd

df = pd.read_csv('./data/my_file (1).csv')
df
```

```python
df1=df[["Artist","Shows","Tour title","Year(s)","Average gross","Peak","Rank"]].copy()
df1
```

```python
df2
```

```python
df2.drop_duplicates()
```

```python
df3=df2.iloc[0:4,0:7].copy()
df3
```

```python
df4=df2.iloc[6:15,0:7].copy()
df4
```

```python
df5=pd.concat([df3,df4])
df5
```

```python
df6=df5.rename(columns={"Tour title":"Tour_title","Year(s)":"Years","Average gross":"Average_gross"}).copy()
df6
```

```python
df7=df6.sort_values(["Rank"],ascending=True).copy()
df7
```

```python
df7.reset_index(drop=True)
```

```python

df7['Average_gross'] = df7['Average_gross'].str.replace('$', '').str.replace(',', '').copy()
df7
```

```python
df7
```

```python
df7.iloc[0:6,0:6].copy()
```

```python
df7.reset_index(drop=True,inplace=True)
df7
```

```python
df7["Rank"] = df7["Artist"].rank() 
df7
```

```python
df7["Rank"] = df7["Rank"].astype(int)
df7
```

```python
df7.to_csv("Singer.csv",index=False)
```

```python
df7["Average_gross"]=df7["Average_gross"].astype(int)
df7
```

```python
df8=df7.groupby('Artist')['Average_gross'].agg(average_mean="mean").copy()
df8
```

```python
df9 =pd.merge(df7,df8,on="Artist",how="outer").copy()
df9
```

```python
df9=df9[["Rank","Artist","Shows","Years","Average_gross","Tour_title","average_mean"]]
df9
```

```python
df10 =df9.rename(columns={"average_mean":"Average_mean"}).copy()
df10
```

```python
df10
```

```python
df11 = df10.sort_values("Years",ascending=False).copy()
df11
```

```python
df11.reset_index(drop=True,inplace=True)
df11
```

```python
df12=df11.groupby('Artist').agg({
    'Shows': 'sum',
    'Average_gross': ['mean'],
    'Rank': ['max','min']
})
df12
```

```python
# Groupby करने के बाद
df12 = df11.groupby('Artist').agg({
    'Shows': 'sum',
    'Average_gross': ['mean'],
    'Rank': ['max','min']
})

# Column names को flatten कर
df12.columns = ['_'.join(col).strip() for col in df12.columns.values]
df12 = df12.reset_index()

df12.columns
df12
```

```python
# अब merge करेगा
df13 = pd.merge(df11, df12, on="Artist", how="outer")

df13
```

```python
df13.reset_index(drop=True,inplace=True)
df13
```

```python
df13.to_csv("Singers.csv",index=False)
```

```python
# present data using matplotlib
import matplotlib.pyplot as plt
import numpy as np

years = [2001,2010,2005,2009,2005,2011]
Sachin = [12,111,200,24,56,78]
Kohli = [100,239,43,12,90,88]
Rohit = [12,111,200,12,90,88]

x = np.arange(len(years))
width = 0.25

plt.bar(x-width,Sachin,width=width,color="blue",label="Sachin")
plt.bar(x,Kohli,width=width,color="orange",label="Kohli")
plt.bar(x+width,Rohit,width=width,color="red",label="Rohit")
plt.legend()
plt.xticks(x,years)
plt.grid(True)
plt.title("Player's Comparison Bar Plot Representation\n")
plt.xlabel("Years  ➡️")
plt.ylabel("Runs  ➡️")
plt.tight_layout()
plt.style.use('petroff10')
#plt.style.available
```

```python
import matplotlib.pyplot as plt 
Fruits = ['Apple','Banana','Grapes']
liked = [500,100,55]
explode =[0.1,0,0,]
colors = ('#47B39C','#FFC154','#f0f2eb')
plt.title("MOST LIKED FRUIT'S")
plt.pie(liked,labels=Fruits,explode=explode,autopct='%1.1f%%',shadow = True,colors=colors, startangle=190,wedgeprops=
    {'edgecolor':"black",'linestyle' :'--'})

plt.tight_layout()
plt.show()
```
