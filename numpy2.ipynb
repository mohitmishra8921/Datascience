# checking memory size of numpy arrays as well as size of python list


```python
import numpy as np
import sys
list1 = list(range(100000))
array = np.array(list1)
print("Size of python list:",sys.getsizeof(list1) *len(list1) ,"bytes")
print("size of numpy arrays:",array.nbytes,"bytes")
```

```python
import numpy as np 
import time 
list1 = list(range(200))
product = np.array(list1)

start = time.time()
result = product *2
end = time.time()
print(result)
print("time taken :", end - start )

```

```python

import time 

list1 = list(range(200))

start = time.time()
result = [x**2 for x in list1]
end = time.time()
print("time taken:",end - start)
```
